```python
# Delivery time
orders['delivery_time'] = (
    orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']
).dt.days

# Delivery delay
orders['delivery_delay'] = (
    orders['order_delivered_customer_date'] - orders['order_estimated_delivery_date']
).dt.days
```

## STEP 3: Data Merging & Cleaning

```python
df = orders.copy()

df = df.merge(customers, on='customer_id', how='left')

df = df.merge(order_items, on='order_id', how='left')

df = df.merge(products, on='product_id', how='left')

df = df.merge(sellers, on='seller_id', how='left')

df = df.merge(payments, on='order_id', how='left')

df = df.merge(reviews, on='order_id', how='left')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()/df.shape[0]*100
```

```python
# Drop product text features (correct column names)
df = df.drop(columns=[
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty"
], errors='ignore')


# Fill product dimensions (very low missing %)
for col in ["product_length_cm","product_height_cm","product_width_cm"]:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())


# Drop missing payment rows (very small % missing ~0.002%)
df = df.dropna(subset=[
    "payment_sequential",
    "payment_type",
    "payment_installments",
    "payment_value"
])


# Drop review text columns (high missing, not useful for analysis)
df = df.drop(columns=[
    'review_comment_title',
    'review_comment_message'
], errors='ignore')
```

```python
df.isnull().sum()/df.shape[0]*100
```
